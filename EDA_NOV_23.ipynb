```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

a = [23,45,56,78,34,67,200,45,78,59]
```

```python
Q1 = np.percentile(a,25,method='midpoint')
Q2 = np.percentile(a,50,method='midpoint')
Q3 = np.percentile(a,75,method='midpoint')
print(Q1)
print(Q2)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
ll = Q1 - 1.5*IQR
ul = Q3 - 1.5*IQR
print(ll)
print(ul)
```

```python
out = []
for i in a:
    if((i<ll) or (i>ul)):
        out.append(i)

print(out)
```
