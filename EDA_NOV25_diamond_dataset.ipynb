```python
Q1 = np.percentile(df['carat'],25,method='midpoint')
Q2 = np.percentile(df['carat'],50,method='midpoint')
Q3 = np.percentile(df['carat'],75,method='midpoint')

print(Q1)
print(Q2)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
upper_limit = Q3+1.5*IQR
lower_limit = Q1-1.5*IQR
print(upper_limit)
print(lower_limit)
```

```python
outlier=[]
for x in df['carat']:
    if((x>upper_limit) or (x<lower_limit)):
        outlier.append(x)
outlier
```

```python
df1 = (df['carat'] > upper_limit ) | (df['carat']<lower_limit)
df1
```

```python
x = df.loc[df1]
x
```

```python
df.drop(x.index,inplace=True)
```
